### несколько графиков на одной координатной сетке

```python
a = np.array(data['Student_ID'])
b = np.array(data['Sleep_Duration'])
c = np.array(data['Study_Hours'])

fig1 = plt.figure()
axs1 = fig1.add_subplot(xlabel='айди студентов', ylabel='кол-во часов', title='график функции')

gph1 = axs1.plot(a, b, marker='o', color='yellow', label='длительность сна')
gph2 = axs1.plot(a, c, marker='o', color='pink', label='длительность учебы')

axs1.grid()
axs1.legend();
```

### fix 

```python
a = np.array(data['Student_ID'][:10])  
b = np.array(data['Sleep_Duration'][:10])
c = np.array(data['Study_Hours'][:10])

fig1 = plt.figure(figsize=(10, 6))
axs1 = fig1.add_subplot(111)
axs1.set_xlabel('ID')
axs1.set_ylabel('часы')
axs1.set_title('Sleep Duration vs Study Hours')

gph1 = axs1.plot(a, b, marker='o', color='yellow', label='Sleep Duration')
gph2 = axs1.plot(a, c, marker='o', color='pink', label='Study Hours')

axs1.grid()
axs1.legend()
plt.show()
```

### обращение к отдельным строкам с помощью условия

```python
data_year1 = data[data['University_Year']=='1st Year']
data_year2 = data[data['University_Year']=='2nd Year']
data_year3 = data[data['University_Year']=='3rd Year']
data_year4 = data[data['University_Year']=='4th Year']
```

```python
data_year_act1 = data_year1[['Physical_Activity']].mean()
data_year_act2 = data_year2[['Physical_Activity']].mean()
data_year_act3 = data_year3[['Physical_Activity']].mean()
data_year_act4 = data_year4[['Physical_Activity']].mean()
```

```python
fig1 = plt.figure()
axs1 = fig1.add_subplot(xlim=[0, 3.5], xticks=range(0, 4), xlabel='год обучения', ylabel='средняя активность')
gph1 = axs1.plot(['1st Year', '2nd Year', '3rd Year', '4th Year'], [data_year_act1, data_year_act2, data_year_act3, data_year_act4 ], marker='s', color="pink")
```

### Шаг 9 - несколько координатных осей на одном графике

```python
fig1 = plt.figure(figsize=(12, 6))
axs1 = fig1.add_subplot(1, 2, 1); scatter = axs1.scatter(data['Screen_Time'], data['Sleep_Quality'], c=data['Caffeine_Intake'], cmap='viridis', alpha=0.7, s=data['Sleep_Duration']*5)
axs1.set_xlabel('Screen Time (hours)'); axs1.set_ylabel('Sleep Quality (1-10)'); axs1.set_title('Screen Time vs Sleep Quality')

axs2 = fig1.add_subplot(1, 2, 2); axs2.scatter(data_male['Physical_Activity'], data_male['Sleep_Duration'], color='blue', alpha=0.5, s=30, label='Male')
axs2.scatter(data_female['Physical_Activity'], data_female['Sleep_Duration'], color='red', alpha=0.5, s=30, label='Female')
axs2.scatter(data_other['Physical_Activity'], data_other['Sleep_Duration'], color='green', alpha=0.5, s=30, label='Other')
axs2.set_xlabel('Physical Activity (minutes)'); axs2.set_ylabel('Sleep Duration (hours)'); axs2.set_title('Physical Activity vs Sleep Duration'); axs2.legend()

plt.tight_layout()
plt.show()
```

### Шаг 12 - круговая диаграмма pie
